```python
from tensorflow.python.summary.summary_iterator import summary_iterator
import pandas as pd

def extract_tb_data(event_file):
    steps = []
    values = []
    # print(summary_iterator(event_file)[0])
    for e in summary_iterator(event_file):
        # print(e)
        for v in e.summary.value:
            if v.tag == 'train/loss':  # replace 'loss' with your specific metric
                steps.append(e.step)
                values.append(v.simple_value)

    return pd.DataFrame({'Step': steps, 'Value': values})

df1 = extract_tb_data('./switch_roasted_trained/switch_roast_code_data_0_1/runs/Dec02_15-00-04_terminator5.cs.rice.edu/events.out.tfevents.1701550804.terminator5.cs.rice.edu.2095288.0')
df2 = extract_tb_data('./switch_roasted_trained/switch_roast_code_data_0_2/n/Dec02_15-01-01_terminator5.cs.rice.edu/events.out.tfevents.1701550861.terminator5.cs.rice.edu.2095898.0')
df3 = extract_tb_data('./switch_roasted_trained/switch_roast_code_data_0_3/runs/Dec02_15-01-37_terminator5.cs.rice.edu/events.out.tfevents.1701550897.terminator5.cs.rice.edu.2096289.0')
df4 = extract_tb_data('./switch_roasted_trained/switch_roast_code_data_0_4/runs/Dec02_15-03-09_terminator5.cs.rice.edu/events.out.tfevents.1701550989.terminator5.cs.rice.edu.2097068.0')
df5 = extract_tb_data('./switch_roasted_trained/switch_roast_code_data_0_5/runs/Dec02_15-03-24_terminator5.cs.rice.edu/events.out.tfevents.1701551004.terminator5.cs.rice.edu.2097331.0')
df6 = extract_tb_data('./switch_trained/switch_code_data/runs/Dec01_20-45-07_terminator5.cs.rice.edu/events.out.tfevents.1701485107.terminator5.cs.rice.edu.1912979.0')
```
